```python
import pandas as pd
import numpy as np
```

```python
folder_path = r"/Users/foramkamdar/Desktop/oehrn_lab/repos/pd_nonmotor/stimuli/blocks/pdnm002_sessions/future sessions"

# load all the csv files in the folder into a multiple dataframes
dfs = []
for i in range(1, 7):
    file_path = f"{folder_path}/session_{i}.csv"
    df = pd.read_csv(file_path)
    dfs.append(df)
```

```python
len(dfs)
```

```python
# work on df[0] first
# work on all dataframes in dfs
for idx, this_df in enumerate(dfs, start=1):

    print(f"\nSession {idx} - balance across val_bin and condition:")
    print(this_df.groupby(['val_bin']).size())

```

```python
# work on df[0] first
# work on all dataframes in dfs
for idx, this_df in enumerate(dfs, start=1):

    print(f"\nSession {idx} - balance across val_bin and condition:")
    print(this_df.groupby(['val_bin', 'condition']).size())

```

```python
# show category balance within each val_bin across sessions
combined = []
for idx, this_df in enumerate(dfs, start=1):
    tmp = this_df.copy()
    tmp['session'] = f'Session {idx}'
    combined.append(tmp)

all_sessions = pd.concat(combined, ignore_index=True)

# keep bins in numeric order (e.g., 1-2, 2-3, ...)
bin_order = sorted(
    all_sessions['val_bin'].dropna().unique(),
    key=lambda x: float(str(x).split('-')[0])
)

for vb in bin_order:
    subset = all_sessions[all_sessions['val_bin'] == vb]
    table = (
        subset.groupby(['Category', 'session'])
        .size()
        .unstack(fill_value=0)
        .sort_index()
    )
    print(f"\nVal_bin {vb} - category counts across sessions:")
    print(table)
```
